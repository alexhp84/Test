```python
import pandas as pd
import numpy as np
import math
from sklearn.preprocessing import MinMaxScaler
from IPython.display import Image
```

<h1>03 Scaling: Min-Max & Standard</h1>

<h3>Min-max normalisation</h3>

```python
mm_data = np.array([18,42,66,90])
min = mm_data.min()
max = mm_data.max()
scaled_mm_data = [(x - min) / (max - min) for x in mm_data]
print('Scaled = ', scaled_mm_data)
```

<h3>Standard deviation</h3>

```python
sd_data = np.array([8, 10, 10, 14, 14, 16])
sd_n = sd_data.size

sd_mean = sum(sd_data) / sd_n

sd_variance = sum((x - sd_mean) ** 2 for x in sd_data) / sd_n
sd_std_deviation = math.sqrt(sd_variance)

for i, x in enumerate(sd_data):
    sd_deviation = x - sd_mean
    sd_scaled_value = sd_deviation / sd_std_deviation

print('Value = ', x)
print('Mean = ', sd_mean)
print('Deviation (x - mean) = ', sd_deviation)
print('Z Value Step 1 = ', '(', x, '-', sd_mean, ') /', sd_std_deviation)
print('Z Value Step 2 = ', sd_deviation, '/', sd_std_deviation)
print('Z Value Final  = ', sd_scaled_value)
```

<h3>Comparing the two scalers</h3>

<h4>Min-Max normalisation</h4>
Advantage - It scales the range between two values - usually 1 and 0 for easy comparison
<bt>Disadvantage -  It takes into account outliers so if values are between 10 and 20 but there is one value of 1 and one value of 100, these are included, skewing the maths.</bt>

<h4>Standardisation</h4>
Advantage - It takes into account outliers using the mean and standard deviation, thus ensuring the outliers have less effect, as it takes into account the whole range and not just the min and max values. 
<br>Disadvantage - It can, theoretically, be any number thus making it suitable only for simple models and not suitable for more complex datasets. It is best used when a boundary is needed, such as 0 and 1 as opposed to unbounded datasets.

<h1>04 Linear Regression</h1>

<h3>Fitted line — predict, residual, MSE, R²</h3>

```python
lr_X = [1, 3, 5, 7]
lr_y = [19, 25, 41, 51]
lr_n = len(lr_X)

lr_mean_X = sum(lr_X) / lr_n
lr_mean_y = sum(lr_y) / lr_n

lr_numerator = 0
lr_denominator = 0

for i in range(lr_n):
    lr_deviation_x = lr_X[i] - lr_mean_X
    lr_deviation_y = lr_y[i] - lr_mean_y
    lr_numerator += lr_deviation_x * lr_deviation_y
    lr_denominator += lr_deviation_x ** 2

lr_slope = lr_numerator / lr_denominator
lr_Y_intercept = lr_mean_y - (lr_slope * lr_mean_X)


lr_target_X = 9

lr_prediction = (lr_slope * lr_target_X) + lr_Y_intercept

print('Slope (m) = ', lr_slope)
print('Intercept (c) = ', lr_Y_intercept)
print('Prediction for X = 9 is Y = ', lr_prediction)
```

```python
lr_X_val = 5
lr_actual_y = 41

lr_prediction = (lr_slope * lr_X_val) + lr_Y_intercept
lr_residual = lr_actual_y - lr_prediction

print('Hours Studied (X) = ', lr_X_val)
print('Actual Score (Y) = ', lr_actual_y)
print('Predicted Score = ', lr_prediction)
print('Residual = ', lr_residual)
```

```python
lr_ssr = 0

for i in range(lr_n):
    lr_x_val = lr_X[i]
    lr_actual_y = lr_y[i]
    lr_predicted_y = (lr_slope * lr_x_val) + lr_Y_intercept
    lr_residual = lr_actual_y - lr_predicted_y
    lr_squared_residual = lr_residual ** 2
    lr_ssr += lr_squared_residual
    
    print('Student Index = ', i + 1)
    print('Hours Studied (x) = ', lr_x_val)
    print('Actual Score (y) = ', lr_actual_y)
    print('Predicted Score (y_hat) = ', lr_predicted_y)
    print('Residual (e) = ', lr_residual)
    print('Squared Residual = ', lr_squared_residual)
    print('-------------------------------------------')

lr_mse = lr_ssr / lr_n
lr_rmse = math.sqrt(lr_mse)

print('Sum of Squared Errors (SSE/SSR) = ', lr_ssr)
print('Mean Squared Error (MSE) = ', lr_mse)
print('Root Mean Squared Error (RMSE) = ', lr_rmse)
```

```python
lr_sst = 0

for i in range(lr_n):
    lr_actual_y = lr_y[i]
    
    lr_squared_total_deviation = (lr_actual_y - lr_mean_y) ** 2
    lr_sst += lr_squared_total_deviation

lr_r2 = 1 - (lr_ssr / lr_sst)

print('Total SST = ', lr_sst)
print('R2 = ', lr_r2)
```

<h4>Meaning of the results</h4>
SST - This shows the total error in the scores that would be made if the scores were guessed by someone
<br>R2 - This shows the accuracy of the method by showing, as a percentage (need to multiply by 100 for this on a scale of 0 to 100) the amount of error reduced using Linear Regression

<h1>05 Polynomial Regression</h1>

<h3>Adjusted R²</h3>

```python
pr_n = 10
pr_p = 3
pr_r2 = 0.94


pr_adjr2 = 1 - (((1 - pr_r2) * (pr_n - 1)) / (pr_n - pr_p - 1))

print('Adjusted R2 = ', pr_adjr2)
```

<h3>Multiple choice — building the polynomial features</h3>
B - [1, X1, X1^2] - a bias of ones (1) then X1, then  X1^2

<h3>What is Adjusted R²?</h3>
While R² changes with every new feature added, Adjusted R² only changes if a new feature actually changes the prediction accuracy of the model. We need this as regular R² includes noise and often results in an overfitting model, which Adjusted R² helps avoid.

<h1>06 Logistic Regression</h1>

<h3>A Sigmoid probability</h3>

```python
sp_B0 = -3
sp_B1 = 0.6
sp_x = 3
e_constant = 2.718

sp_z = sp_B0 + (sp_B1 * sp_x)
sp_exponent = e_constant ** (-sp_z)
sp_probability = 1 / (1 + sp_exponent)

print('Linear Log-Odds (z) = ', sp_z)
print('Sigmoid Calculation =  ', sp_exponent)
print('Final Probability   = ', sp_probability)
```

<h3>B Sigmoid probability</h3>

```python
sp_B0 = -3
sp_B1 = 0.6
sp_x = 8
e_constant = 2.718

sp_z = sp_B0 + (sp_B1 * sp_x)
sp_exponent = e_constant ** (-sp_z)
sp_probability = 1 / (1 + sp_exponent)

print('Linear Log-Odds (z) = ', sp_z)
print('Sigmoid Calculation =  ', sp_exponent)
print('Final Probability   = ', sp_probability)
```

<h3>★ Bonus Question ★</h3>

```python
sp_probability2 = 0.8

sp_odds = sp_probability2 / (1- sp_probability2)
sp_log_odds = math.log(sp_odds)
sp_x = (sp_log_odds - sp_B0) / sp_B1
print('The number of words needed for an email to be marked as spam with an 80% accuracy is', sp_x)
```
